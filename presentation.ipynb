- Every state looks like [Cognitive Index] - [Behavioural Index]

## Reachability via Monte Carlo Simulations 
- Reachbility difficult to determine analytically
- Monte Carlo Simulations $\rightarrow$ repeated random sampling to approximate probability distribution
